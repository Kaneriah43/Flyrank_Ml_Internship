## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

*My lane is Performance-based opportunity modeling. The task is to assign each content page a priority score and rank pages from highest to lowest opportunity, so a content team knows exactly where to spend their limited editorial hours first.*

*This is not classification, I am not simply labelling pages as "needs refresh" or "doesn't need refresh." A binary yes/no answer throws away the ordering information that makes the output actionable. A content team can only work on a few pages per week; they need a ranked queue, not a list of flags.*

*This is not clustering — I am not grouping similar pages together. I am scoring individual pages against a business outcome.*

This is scoring/ranking because:

* Every page gets a numeric score (higher = higher priority)
* Pages are ranked by that score into an actionable queue
* The output directly maps to a real decision: "work on page 1 before page 2"

*The parallel to my dissertation is direct — fraud detection is also a scoring problem. The model does not just say "fraud or not fraud." It assigns a risk score to every transaction, and analysts review the highest-scored ones first. Same structure, different domain.*

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

*Target: is_declining_label (proxy, rule-defined)*

*There is no ground-truth label in this dataset — no column that says "this page needed a refresh." So I construct a proxy: a page is a priority candidate if it is trending down AND has above-median impressions — still visible to users but losing clicks.*

*The label comes from a defined rule, not an observed outcome. This is standard practice when real-world results take months to measure.*

*This proxy is an assumption — it is my best available signal for "worth fixing," not a confirmed outcome.*

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv")

df["is_declining_label"] = (
    (df["trend_direction"] == "down") &
    (df["impressions_90d"] > df["impressions_90d"].median())
).astype(int)

df[["content_id", "content_type", "impressions_90d",
    "ctr", "trend_direction", "position_tier",
    "is_declining_label"]].head(10)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

*The opportunity signal is not one thing — it is a combination. A page needs refresh when it has high impressions and declining trend and poor CTR relative to its position tier and has not been updated recently. Any one of those alone is not enough.*

*An if-statement can check one condition at a time. It cannot learn that avg_position = 3 with ctr = 0.01 is a crisis, while avg_position = 9 with ctr = 0.01 is normal. That context-dependence is exactly what a learned model captures — and exactly what makes the pattern too tangled for a fixed rule to follow.*
