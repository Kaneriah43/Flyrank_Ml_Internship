## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

*Decision: Which pages should the content team prioritise for refresh or optimisation this week?*

*Who acts on it: The content strategist or SEO manager who decides where to spend limited editorial hours. They get a ranked queue and act on the top items first.*

*Cost of a wrong call:*

*   False positive (flagging a page that didn't need work) → editorial time wasted on content that was already performing fine — opportunity cost.
*   False negative (missing a page that genuinely needed refresh) → that page keeps losing clicks and ranking positions, compounding week over week — revenue cost.

*The asymmetry matters: missing a declining page hurts more than over-flagging one, because search ranking losses are slow to recover. So the model should be tuned to catch the misses, not just be "accurate" on average — exactly the same precision-recall tradeoff I face in fraud detection, where a missed fraud is far costlier than a false alarm.*

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv")
```

```python
# --- Number 1: How much performance spread exists across pages? ---
print("OPPORTUNITY SPREAD")
print(f"Top 10% CTR:    {df['ctr'].quantile(0.90):.3f}")
print(f"Median CTR:     {df['ctr'].quantile(0.50):.3f}")
print(f"Bottom 10% CTR: {df['ctr'].quantile(0.10):.3f}")
print(f"→ Top pages get {df['ctr'].quantile(0.90) / df['ctr'].quantile(0.10):.1f}x more clicks than bottom pages at similar impressions")
```

```python
# --- Number 2: How many pages are declining and getting clicks? ---
print("\nDECLINING PAGES THAT STILL HAVE IMPRESSIONS")
declining = df[(df["trend_direction"] == "down") & (df["impressions_90d"] > df["impressions_90d"].median())]
print(f"Pages trending down but above median impressions: {len(declining):,}")
print(f"That is {len(declining)/len(df)*100:.1f}% of all pages — these are the priority targets")
```

```python
# --- Number 3: Does position tier actually predict CTR? ---
print("\nPOSITION TIER vs CTR")
pt = df.groupby("position_tier")["ctr"].mean().round(4)
print(pt.to_string())
print(f"→ CTR gap between best and worst tier: {pt.max() - pt.min():.4f} — a real, actionable signal")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

*What my work can say:*

* "Pages with these characteristics were observed to have lower CTR at the same position tier"
* "The model ranked these pages as higher priority for refresh based on measured signals"
* "Historically, pages matching this pattern showed declining trend direction"
* "This queue is a decision-support tool — it surfaces candidates, a human decides"

*What my work CANNOT say:*

* "Refreshing this page will improve its Google ranking" — I have no causal proof, only correlation
* "I am predicting Google's algorithm" — I am modelling observed performance patterns, not reverse-engineering ranking logic
* "This page will get X% more clicks if updated" — future performance depends on factors outside the data
* "This finding applies to all websites" — the data is one anonymized client slice, not a universal sample*
